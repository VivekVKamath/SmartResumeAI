## Objective

This notebook implements a runnable baseline for an AI-powered Resume Analyzer.

The application performs the following tasks:

- Uploads a Resume (PDF)
- Extracts Resume Text
- Accepts a Job Description
- Compares Resume Skills with Job Requirements
- Calculates ATS Compatibility Score
- Displays Matching Skills
- Displays Missing Skills
- Generates Resume Improvement Suggestions
- Generates Interview Questions

This notebook serves as the baseline implementation for the semester capstone project.

```python
## Import Required Libraries

from pypdf import PdfReader
from google.colab import files
from IPython.display import display

import ipywidgets as widgets

from reportlab.pdfgen import canvas
from docx import Document

import re
import os
```

```python
## Upload Resume PDF

uploaded = files.upload()
```

```python
## Read Resume

filename = list(uploaded.keys())[0]

reader = PdfReader(filename)

resume_text = ""

for page in reader.pages:

    text = page.extract_text()

    if text:

        resume_text += text

print("Resume Loaded Successfully!")
```

```python
## Preview Resume

print(resume_text[:1500])
```

```python
## Enter Job Description

job_description = """
Software Engineer Intern

Required Skills

Python
Java
SQL
Git
Docker
AWS
REST API
React
Problem Solving
Communication
Teamwork

Responsibilities

Develop Web Applications

Build REST APIs

Work with SQL Databases

Deploy Applications

Collaborate with Team Members
"""
```

```python
## Skills Database

skills = [

"python",
"java",
"c",
"c++",
"javascript",
"typescript",
"react",
"angular",
"node",
"express",
"sql",
"mysql",
"postgresql",
"mongodb",
"firebase",

"html",
"css",

"docker",
"kubernetes",

"aws",
"azure",
"gcp",

"git",

"linux",

"machine learning",
"deep learning",
"tensorflow",
"pytorch",

"opencv",

"numpy",
"pandas",

"flask",
"django",
"fastapi",

"php",

"flutter",
"dart",

"unity",
"unreal",

"rest api",
"api",

"problem solving",
"communication",
"teamwork",
"leadership"

]
```

```python
## Extract Skills

resume_lower = resume_text.lower()

job_lower = job_description.lower()

resume_skills = set()

job_skills = set()

for skill in skills:

    if skill in resume_lower:

        resume_skills.add(skill)

    if skill in job_lower:

        job_skills.add(skill)
```

```python
## Compare Skills

matched_skills = sorted(

    list(

        resume_skills.intersection(job_skills)

    )

)

missing_skills = sorted(

    list(

        job_skills - resume_skills

    )

)

extra_skills = sorted(

    list(

        resume_skills - job_skills

    )

)
```

```python
## Calculate ATS Score

if len(job_skills) == 0:

    ats_score = 0

else:

    ats_score = round(

        len(matched_skills)

        /

        len(job_skills)

        *

        100

    )

print("="*60)

print("ATS SCORE")

print("="*60)

print(f"{ats_score}%")
```

```python
print("\nMatched Skills")

print("-"*40)

for skill in matched_skills:

    print("✓", skill)

print("\nMissing Skills")

print("-"*40)

for skill in missing_skills:

    print("✗", skill)

print("\nExtra Skills")

print("-"*40)

for skill in extra_skills:

    print("•", skill)
```

```python
# Step 10: Resume Summary - Generate a summary of the uploaded resume based on the extracted skills.

summary = f"""
The uploaded resume contains {len(resume_skills)} recognized technical/professional skills.

The candidate matches {len(matched_skills)} out of {len(job_skills)} required job skills.

Overall ATS Compatibility Score: {ats_score}%

The resume demonstrates knowledge in:

{', '.join(sorted(resume_skills)) if resume_skills else "No recognized skills found."}

The resume should be improved by addressing the missing skills listed below.
"""

print(summary)
```

```python
# Step 11: Resume Improvement Suggestions

suggestions = []

for skill in missing_skills:
    suggestions.append(f"Add projects or experience demonstrating {skill.title()}.")

suggestions.extend([
    "Use measurable achievements (e.g., Improved system performance by 20%).",
    "Use action verbs such as Developed, Designed, Implemented, Optimized.",
    "Tailor your resume for every job application.",
    "Keep technical skills near the top of the resume.",
    "Include GitHub or portfolio links if available."
])

print("="*60)
print("RESUME IMPROVEMENT SUGGESTIONS")
print("="*60)

for i, suggestion in enumerate(suggestions, start=1):
    print(f"{i}. {suggestion}")
```

```python
# Step 12: Generate Interview Questions

interview_questions = []

for skill in matched_skills:
    interview_questions.append(
        f"Explain a project where you used {skill.title()}."
    )

for skill in missing_skills:
    interview_questions.append(
        f"What do you know about {skill.title()} and how would you learn it?"
    )

if len(interview_questions) < 10:
    interview_questions.extend([
        "Describe a challenging debugging experience.",
        "Explain object-oriented programming.",
        "How do you work in a team environment?"
    ])

print("="*60)
print("SUGGESTED INTERVIEW QUESTIONS")
print("="*60)

for i, question in enumerate(interview_questions, start=1):
    print(f"{i}. {question}")
```

```python
# Step 13: Resume Statistics

print("="*60)
print("RESUME STATISTICS")
print("="*60)

print("Total Resume Skills :", len(resume_skills))
print("Required Job Skills :", len(job_skills))
print("Matched Skills      :", len(matched_skills))
print("Missing Skills      :", len(missing_skills))
print("Extra Skills        :", len(extra_skills))
print("ATS Score           :", f"{ats_score}%")
```

```python
# Step 14: Generate Professional Report

report = f"""
============================================================
                SMART RESUME AI REPORT
============================================================

RESUME SUMMARY
--------------
{summary}

============================================================

ATS SCORE
---------
{ats_score}%

============================================================

MATCHED SKILLS
--------------
{chr(10).join('✓ ' + skill for skill in matched_skills)}

============================================================

MISSING SKILLS
--------------
{chr(10).join('✗ ' + skill for skill in missing_skills)}

============================================================

EXTRA SKILLS
------------
{chr(10).join('• ' + skill for skill in extra_skills)}

============================================================

RESUME IMPROVEMENT SUGGESTIONS
------------------------------
{chr(10).join(str(i+1)+". "+s for i,s in enumerate(suggestions))}

============================================================

INTERVIEW QUESTIONS
-------------------
{chr(10).join(str(i+1)+". "+q for i,q in enumerate(interview_questions))}

============================================================
END OF REPORT
============================================================
"""
```

```python
# Step 15: Display Final Report

print(report)
```

```python
# Step 16: Save Report in Memory - The report is stored in the variable `report` and will be used in Part 3 to generate downloadable files.

print("Report generated successfully!")

print("\nLength of Report:", len(report), "characters")
```

# Step 17: Download Report

Choose a format and download the generated report.

Supported Formats

- TXT
- PDF
- DOCX
- HTML

```python
from google.colab import files
from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas
from docx import Document
import ipywidgets as widgets
from IPython.display import display
```

```python
## Select Report Format

format_dropdown = widgets.Dropdown(
    options=[
        ("TXT File", "txt"),
        ("PDF File", "pdf"),
        ("DOCX File", "docx"),
        ("HTML File", "html")
    ],
    value="txt",
    description="Format:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="300px")
)

download_button = widgets.Button(
    description="Generate & Download",
    button_style="success",
    icon="download"
)

display(format_dropdown)
display(download_button)
```

```python
## Generate Selected Report

def generate_report(button):

    selected = format_dropdown.value

    if selected == "txt":

        filename = "SmartResumeAI_Report.txt"

        with open(filename, "w", encoding="utf-8") as f:
            f.write(report)

        files.download(filename)

        return


    elif selected == "pdf":

        filename = "SmartResumeAI_Report.pdf"

        pdf = canvas.Canvas(filename, pagesize=letter)

        width, height = letter

        y = height - 40

        pdf.setFont("Helvetica", 10)

        for line in report.split("\n"):

            pdf.drawString(30, y, line)

            y -= 15

            if y < 40:
                pdf.showPage()
                pdf.setFont("Helvetica", 10)
                y = height - 40

        pdf.save()

        files.download(filename)

        return


    elif selected == "docx":

        filename = "SmartResumeAI_Report.docx"

        doc = Document()

        doc.add_heading("SmartResumeAI Report", level=1)

        doc.add_paragraph(report)

        doc.save(filename)

        files.download(filename)

        return


    elif selected == "html":

        filename = "SmartResumeAI_Report.html"

        html = f"""
        <html>

        <head>

        <title>SmartResumeAI Report</title>

        <style>

        body {{
            font-family: Arial;
            margin:40px;
            background:#f5f5f5;
        }}

        pre {{
            background:white;
            padding:20px;
            border-radius:10px;
            font-size:15px;
        }}

        h1 {{
            color:#003366;
        }}

        </style>

        </head>

        <body>

        <h1>SmartResumeAI Report</h1>

        <pre>{report}</pre>

        </body>

        </html>
        """

        with open(filename, "w", encoding="utf-8") as f:

            f.write(html)

        files.download(filename)

        return
```

```python
download_button.on_click(generate_report)
```
